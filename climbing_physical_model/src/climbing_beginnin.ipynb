Import packages


```python
#https://adamheins.com/blog/lagrangian-mechanics-three-ways#5
import numpy as np
import math
from scipy.integrate import odeint
import sympy as sym
from sympy import diff
from sympy.matrices import Matrix
from sympy.physics.mechanics import dynamicsymbols, init_vprinting
import matplotlib.pyplot as plt
```

```python
init_vprinting()
```

```python
#t, m, q_1, q_2, I, L, g = sym.symbols('t, m, q_1, q_2, I, L, g')
t, m, L, g = sym.symbols('t, m, L, g')
```

```python
#q_1, q_2 = dynamicsymbols('q_1, q_2') 
#q_1 = q_1(t)
#q_2 = q_2(t)
q_1, q_2 = dynamicsymbols('q_1, q_2') 
omega_1, omega_2 = dynamicsymbols ('omega_1, omega_2')
#q1_dot = diff(q1)
```

```python
q_1_d = diff(q_1)
q_2_d = diff(q_2)
q_1_dd = diff(q_1_d)
q_2_dd = diff(q_2_d)
```

```python
x_0 = 0
y_0 = 0
```

Direct Dynamic

```python

x_B = x_0 + 2*L*sym.cos(q_2)
y_B = y_0 + 2*L*sym.sin(q_2)
x_C = 2*L*sym.cos(q_2) + L*sym.cos(2*q_1-q_2)
y_C = 2*L*sym.sin(q_2)  + L*sym.sin(2*q_1-q_2)
```

```python
rB = sym.Matrix([[x_B, y_B, 0]])
rC = sym.Matrix([[x_C, y_C, 0]])
r1 = rB/2
r2 = r1 + rC/2
r3 = rC/2
v1 = sym.diff(r1,t)
v2 = sym.diff(r2,t)
v3 = sym.diff(r3,t)
omega_1 = sym.Matrix([[0, 0, q_1_d]])
omega_2 = sym.Matrix([[0, 0, q_2_d]])
```

Kinetic and Potential Energy

```python
I = sym.Rational(1,3)*m*2*L**2
T1 = sym.Rational(1/2)*I*omega_1.dot(omega_1) + sym.Rational(1/2)*m*v1.dot(v1) 
T2 = sym.Rational(1/2)*I*q_2_d**2 + sym.Rational(1/2)*m*v2.dot(v2) 
T3 = sym.Rational(1/2)*m*v3.dot(v3) 
T = sym.simplify(T1 + T2 + T3)
T
```

* Potential energy $V$ comes from both the gravitational potential energy of the masses $mgy$ 

```python
G1 = sym.Matrix([[0, -m*g, 0]])
G2 = sym.Matrix([[0, -m*g, 0]])
V1 = G1.dot(r1)
V2 = G2.dot(r2)
V3 = G2.dot(r3)
V = V1 + V2
V
```

* Setting the Lagragian

```python
L = T - V

L.simplify()
```

```python
LE_q_1 = sym.diff(L, q_1_d)
LE_q_1.simplify()
```

```python
LE_q_2 = sym.diff(L, q_2_d)
LE_q_2.simplify()
```
