```python
import numpy as np
from sympy import *
```

```python
#x, y = symbols('x y')
x,y,z=symbols('x y z')
```

```python
sym_matrix = Matrix([
    [x + y, x * y],
    [x - y, x / 2]
])
```

```python
sym_matrix
```

```python
f_num = lambdify((x, y), sym_matrix, 'numpy')
```

```python
result = f_num(2, 3)
print(result)
```
